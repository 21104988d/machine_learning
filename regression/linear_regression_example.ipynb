```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('Hyderabad.csv')
data.head()
```

```python
plt.scatter(x = data['Area'], y = data['Price'])
plt.xlabel('Area')
plt.ylabel('Price')
plt.title('House price per area')
plt.show()
```

```python
linear = LinearRegression()
linear.fit(data[['Area']], data['Price'])
print("Coefficient is : ", round(float(linear.coef_[0]), 2), "\nIntercept is : ", round(float(linear.intercept_), 2))
plt.scatter(x = data['Area'], y = data['Price'])
plt.plot(data['Area'], linear.predict(data[['Area']]), color = 'red')
plt.xlabel('Area')
plt.ylabel('Price')
plt.title('House price per area')
plt.show()
```
